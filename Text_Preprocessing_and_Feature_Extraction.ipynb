```python

import nltk
import re
import pandas as pd

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

# Step 1: Sample Text
text = "Machine Learning is AMAZING! Students are learning Python and NLP techniques in 2026."

print("Original Text:")
print(text)

# Step 2: Convert to lowercase
text = text.lower()
print("\nLowercase Text:")
print(text)

# Step 3: Remove punctuation and numbers
text = re.sub(r'[^a-z\s]', '', text)
print("\nCleaned Text:")
print(text)

# Step 4: Tokenization
tokens = word_tokenize(text)
print("\nTokens:")
print(tokens)

# Step 5: Stopword Removal
stop_words = set(stopwords.words('english'))
filtered_tokens = [
    word for word in tokens if word not in stop_words
]
print("\nAfter Stopword Removal:")
print(filtered_tokens)

# Step 6: Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]
print("\nAfter Stemming:")
print(stemmed_words)

# Step 7: Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(word) for word in filtered_tokens
]
print("\nAfter Lemmatization:")
print(lemmatized_words)

# Step 8: Feature Extraction
documents = [
    "Machine learning is interesting",
    "Python is used for machine learning",
    "Natural language processing uses Python"
]

# Bag of Words
cv = CountVectorizer()
bow = cv.fit_transform(documents)

print("\nBag of Words:")
print(pd.DataFrame(
    bow.toarray(),
    columns=cv.get_feature_names_out()
))

# TF-IDF
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(documents)

print("\nTF-IDF Matrix:")
print(pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
).round(3))
```

```python

import re
import pandas as pd
from sklearn.feature_extraction.text import (
    CountVectorizer, TfidfVectorizer, ENGLISH_STOP_WORDS
)

# Sample dataset
documents = [
    "Machine Learning is very useful!",
    "Students are learning Machine Learning.",
    "Python is used for Data Science.",
    "Data Science and Machine Learning use Python."
]

# Step 1: Text Preprocessing
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    words = text.split()
    words = [w for w in words if w not in ENGLISH_STOP_WORDS]
    return " ".join(words)

cleaned_docs = [preprocess(doc) for doc in documents]

print("Original Documents:")
for doc in documents:
    print(doc)

print("\nPreprocessed Documents:")
for doc in cleaned_docs:
    print(doc)

# Step 2: Unigram Extraction
unigram = CountVectorizer(ngram_range=(1, 1))
unigram_matrix = unigram.fit_transform(cleaned_docs)

print("\nUnigram Features:")
print(unigram.get_feature_names_out())

# Step 3: Bigram Extraction
bigram = CountVectorizer(ngram_range=(2, 2))
bigram_matrix = bigram.fit_transform(cleaned_docs)

print("\nBigram Features:")
print(bigram.get_feature_names_out())

# Step 4: Trigram Extraction
trigram = CountVectorizer(ngram_range=(3, 3))
trigram_matrix = trigram.fit_transform(cleaned_docs)

print("\nTrigram Features:")
print(trigram.get_feature_names_out())

# Step 5: TF-IDF Feature Extraction
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(cleaned_docs)

df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

print("\nTF-IDF Feature Matrix:")
print(df.round(3))
```
